# Selecionando colunas e filtrando linhas

```python
import pandas as pd
```

## Selecionando colunas

```python
data = {'Nome': ['Adriano', 'Rodrigo', 'Juliano', 'Mateus', 'Luiza', 'Rodrigo', 'Ayres', 'Aninha'],
		'Sobrenome': ['Soares', 'Tadewald', 'Faccioni', 'Kienzle', 'Cherobini', 'Vanzelotti', 'Júnior', 'Tadewald'],
		'Idade Atual': [31, 31, 25, 35, 22, 21, 28, 20],
		}

df = pd.DataFrame(data)
df.Sobrenome
```

```python
colunas = [col for col in df.columns if not col in ['Sobrenome']]
df[colunas]
```

## Filtrando por indexação (.iloc)

```python
#seleção da linha 1 a 3 (posição vetorial)
df.iloc[1:3]
```

```python
#seleção da linha 3 em diante, pegando a coluna 0
df.iloc[3:, 0]
```

```python
#seleção da linha 3 em diante, pegando a coluna 0 e 2
df.iloc[3:, 0:1]
```

## Filtrando por condições (.loc)

```python
df[df['Nome'] == 'Rodrigo']
```

```python
df.loc[df['Nome'] == 'Rodrigo', ['Sobrenome']]
```

## Filtrando por múltiplas condições

```python
df.loc[((df['Nome'] == 'Rodrigo') & (df['Sobrenome'] == 'Tadewald')), ['Sobrenome']]
```

```python
df.loc[((df['Nome'] == 'Rodrigo') | (df['Nome'] == 'Juliano')), ['Sobrenome', 'Idade Atual']]
```

## Filtrando por lista de valores (.isin)

```python
nomes_filtro = ['Rodrigo', 'Juliano']
df.loc[df['Nome'].isin(nomes_filtro)]
```

```python
df['Nome'].str.contains('Adri')
```

```python
df.loc[df['Nome'].str.contains('Adri')]
```
