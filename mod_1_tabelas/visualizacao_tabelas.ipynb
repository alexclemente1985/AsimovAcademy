## Estilizações

### Formatando floats

```python
df_tempo = pd.DataFrame(np.random.rand(10,2)*5,
                        index=pd.date_range(start="2023-01-01", periods=10),
                        columns=["Porto Alegre", "São Paulo"])

df_tempo.style.format(decimal=',', precision=3)
```

### Estilizando colunas (ex: 'caixa alta' no nome das colunas)

```python
df_tempo.style.format_index(str.upper, axis=1)
```

### Estilizando index (ex: formatação da data)

```python
formatar_data = lambda v: v.strftime('%d/%m/%Y')

df_tempo.style.format_index(formatar_data, axis=0)
```

### Adicionando título no dataframe

```python
df_tempo.style.set_caption('Condições do tempo')
```

### Adicionando gradiente (formatação condicional)

```python
## vmax-> qual valor vai receber cor mais forte (quanto mais próximo ao valor, mais forte a cor)
## vmin-> inverso do vmax
## cmap-: por meio da matplotlib, permite criar um mapa de cores baseado nos valores vmin, vmax e os da tabela

df_tempo.style.background_gradient(axis=None, vmin=1, vmax=5, cmap="YlGnBu")
```

### Modificando valores

```python
def condicao_chuva(v):
    if v > 2.75:
        return 'Chuva forte'
    elif v < 1.75:
        return 'Seco'
    else:
        return 'Chuva normal'

df_tempo.style.format(condicao_chuva)
```

### Juntando tudo o que foi passado anteriormente (criação de função de configuração do style)

```python
def embelezar(styler):
    styler.set_caption('Condição do Tempo (função EMBELEZAR)')
    styler.format(condicao_chuva)
    styler.format_index(formatar_data)
    styler.background_gradient(axis=None, vmin=1, vmax=5, cmap="YlOrRd")

    return styler

df_tempo.style.pipe(embelezar)
```
