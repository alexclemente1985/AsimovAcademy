# Importando arquivos EXCEL

## Lendo arquivos .xlsx

```python
import pandas as pd
from pathlib import Path

#Acrescentar o '..' para pegar pasta datasets global
dataset_vendas = Path.joinpath(Path.cwd(),'..', 'datasets', 'vendas.xlsx')

#id_venda vira índice do dataframe; decimais são definidos com vírgula; uso das três primeiras colunas
df = pd.read_excel(dataset_vendas, index_col=1, decimal=',', usecols=[0,1,2])
df
```

## Lendo arquivos .csv

```python
dataset_vendas = Path.joinpath(Path.cwd(),'..', 'datasets', 'vendas.csv')

df = pd.read_csv(dataset_vendas, index_col=0, sep=';', decimal=',')
df
```
