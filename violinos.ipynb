```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from pathlib import Path
import re
import numpy as np
```

```python
def format_percent(value):
    return f'{100*value:.2f}%'

def remove_html(html):
    pattern = re.compile('<.*?>')
    cleantext = re.sub(pattern, '', html)
    return cleantext.replace('*', '').strip()
```

```python
dataset = Path.joinpath(Path.cwd(), 'dataset')
srp_path = Path.joinpath(dataset, 'survey_results_public.csv')
df = pd.read_csv(srp_path, index_col='ResponseId')
df
```

```python
sns.set()
sns.set_context('notebook')
```

```python
quantile_args = {
    'linewidth': 2,
    'color': '#453698',
    'linestyle': '--',
}
mean_args = {
    'linewidth': 1,
    'color': '#342345',
    'linestyle': '-',
}

fig, (ax_cima, ax_baixo) = plt.subplots(figsize=(12, 8), nrows=2, sharex=True)

sns.histplot(data=df, ax=ax_cima, x='WorkExp', binwidth=1, kde=True)

#Verifica onde pegam as linhas dos quartis
for q in [0.25, 0.5, 0.75]:
    ax_cima.axvline(df['WorkExp'].quantile(q), **quantile_args)

#Verifica onde pega a linha da média
ax_cima.axvline(df['WorkExp'].mean(), **mean_args)

sns.boxplot(data=df, ax=ax_baixo, x='WorkExp', whis=3.0)
ax_baixo.axvline(df['WorkExp'].mean(), **mean_args)

fig.suptitle('Anos de experiência programando')

plt.show()
```

```python
#Violin plot -> juntará os dois gráficos acima em um só
quantile_args = {
    'linewidth': 2,
    'color': '#453698',
    'linestyle': '--',
}
mean_args = {
    'linewidth': 1,
    'color': '#342345',
    'linestyle': '-',
}

fig, ax = plt.subplots(figsize=(12, 8))

sns.violinplot(data=df, ax=ax, x='WorkExp')
for q in [0.25, 0.5, 0.75]:
    ax.axvline(df['WorkExp'].quantile(q), **quantile_args)
ax.axvline(df['WorkExp'].mean(), **mean_args)

fig.suptitle('Anos de experiência programando')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 8))

sns.violinplot(data=df, ax=ax, x='WorkExp', y='RemoteWork')
fig.suptitle('Anos de experiência programando')

plt.show()
```
