# Processamento de dados

## Leitura tabela excel

```python
import pandas as pd
from pathlib import Path

dataset = Path.joinpath(Path('__file__').parent, 'dataset','dados.xlsx')
dados_originais = pd.read_excel(dataset)

dados_originais
```

## Filtrando por mês específico

```python
dados_originais['Data/Hora'].apply(lambda dt: (dt.year, dt.month)).unique()

mes_ref = '2023-01'
filtro = dados_originais['Data/Hora'].apply(
    lambda dt: dt.strftime('%Y-%m') == mes_ref
)

dados_filtrados = dados_originais.loc[filtro]

dados_filtrados
```

```python
dados_vendas = dados_filtrados.pivot_table(
    index="Vendedor",
    columns="Produto",
    values="Quantidade",
    aggfunc="sum"
)

dados_vendas
```

```python
dados_vendas = dados_filtrados.pivot_table(
    index="Vendedor",
    columns="Produto",
    values="Quantidade",
    aggfunc="sum",
    margins=True
)

dados_vendas
```

```python
dados_vendas = dados_filtrados.pivot_table(
    index="Vendedor",
    columns="Produto",
    values="Quantidade",
    aggfunc="sum",
    margins=True,
    margins_name="Total"
).sort_values(by="Total")

dados_vendas
```

```python
dados_volume = dados_filtrados.pivot_table(
    index="Vendedor",
    columns="Produto",
    values="Valor Venda",
    aggfunc="sum",
    margins=True,
    margins_name="Total"
).sort_values(by="Total")

dados_volume
```
