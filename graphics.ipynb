```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from pathlib import Path
import re
import numpy as np
```

```python
def format_percent(value):
    return f'{100*value:.2f}%'
    
def remove_html(html):
    pattern = re.compile('<.*?>')
    cleantext = re.sub(pattern, '', html)
    return cleantext.replace('*', '').strip()
```

```python
#Carregamento dos dados e análise prévia
    
dataset = Path.joinpath(Path.cwd(), 'dataset')
srp_path = Path.joinpath(dataset, 'survey_results_public.csv')
srs_path = Path.joinpath(dataset, 'survey_results_schema.csv')
df = pd.read_csv(srp_path, index_col='ResponseId')
    
df_questions = pd.read_csv(srs_path, index_col='qid')

df_questions
```

```python
#Boxplot

sns.set_theme()
sns.set_context('poster')

df['WorkExp'].describe().round(2)
```

```python
##Criação do gráfico
fig, ax = plt.subplots(figsize=(22,8))
sns.boxplot(data=df, x="WorkExp")
fig.suptitle("Anos de experiência programando")

plt.show() ###NOTA: Outliers são representados pelas "bolinhas" ou traços além do quarto quartil
```

```python
#### Aumento da cerca de Tucker (tolerancia para definição de outliers)
fig, ax = plt.subplots(figsize=(22,8))
#####whis -> define a região a partir da qual se consideram os outliers (define a largura da região que define os elementos não outliers)
sns.boxplot(data=df, x="WorkExp", whis=3.0) ##### Padrão geralmente é 1.5
fig.suptitle("Anos de experiência programando (remoção de outliers)")

plt.show()
```

```python
x_noise = np.random.uniform(-0.1,0.1,len(df))
x_scatter = df['WorkExp'] + x_noise
y_scatter = np.random.uniform(-0.3, 0.3, len(df))

fig,ax = plt.subplots(figsize=(22,8))

sns.boxplot(data=df, ax=ax, x='WorkExp', whis=3.0)
sns.scatterplot(data=df, ax=ax, x=x_scatter, y=y_scatter, alpha=0.8, s=20)
fig.suptitle('Anos de experiência programando')

plt.show()
```

```python
#Histograma

fig,ax = plt.subplots(figsize=(12,5))
sns.histplot(data=df, ax=ax, x='WorkExp')
fig.suptitle('Anos de Experiência Trabalhando')

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(12,5))
sns.histplot(data=df, ax=ax, x='WorkExp', binwidth=1.0)
fig.suptitle('Anos de Experiência Trabalhando')

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(12,5))
#Kernel Density Estimative -> KDE (estimativa da densidades dos pontos... desenha uma curva que facilita a visualização)
sns.histplot(data=df, ax=ax, x='WorkExp', binwidth=1.0, kde=True)
fig.suptitle('Anos de Experiência Trabalhando')

plt.show()
```

```python
valores = np.random.normal(size=10_000)

fig,ax = plt.subplots(figsize=(12,5))
#KDE permite visualizar se os dados seguem a distribuição normal
sns.histplot(valores, kde=True)
fig.suptitle('Distribuição Normal')

plt.show()
```
